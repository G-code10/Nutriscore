```python
import psycopg2
import pandas as pd
```

```python
conn = psycopg2.connect(database = "mediatech", 
                        user = "postgres", 
                        host= 'localhost',
                        password = "postgres",
                        port = 5432)
if not conn:
    print("Erreur de connection...")
else:
    print("Connection OK...")
```

```python
conn.rollback()
cur = conn.cursor()

year_1950 = 1950
year_2000 = 2000

# cur.execute("SELECT titre, annee FROM livres ORDER BY annee DESC LIMIT 10;")
cur.execute("""SELECT
        l.titre,
        l.annee,
        CASE
            WHEN l.annee < %s THEN 'avant 1950'
            WHEN l.annee >= %s AND l.annee < %s THEN 'entre 1950 et 2000'
            WHEN l.annee >= %s THEN 'apres 2000'
        END AS epoque
    FROM livres l;""", (year_1950, year_1950, year_2000, year_2000, ))

rows = cur.fetchall()
conn.commit()
# conn.close()
for row in rows:
    print(row)

```

```python
adherents_df = pd.read_sql("SELECT * FROM adherents", conn)
auteurs_df = pd.read_sql("SELECT * FROM auteurs", conn)
emprunts_df = pd.read_sql("SELECT * FROM emprunts", conn)
livres_df = pd.read_sql("SELECT * FROM livres", conn)

print(adherents_df)
print(auteurs_df)
print(emprunts_df)
print(livres_df)
```
