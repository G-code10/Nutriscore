```python
import cv2
import zxingcpp

imgs = ["cb1.jpg", "cb2.jpg", "cb3.jpg", "cb4.jpg", "cb5.jpg", "cb6.jpg", "cb7.jpg", "cb8.jpg", "cb9.jpg", "cb10.jpg"]

for i in imgs:
    img = cv2.imread(i)

    barcodes = zxingcpp.read_barcodes(img, try_rotate=True, try_invert=True, try_downscale=True)
    if barcodes:
        for barcode in barcodes:
            print(i)
            print(barcode.text)
            print(barcode.format)
    else:
        print(f"{i}: Error reading barcode...")
```
